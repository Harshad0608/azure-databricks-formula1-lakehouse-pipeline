```python
#define source file and table name
source_file = f"{landing_folder_path}/{v_batch_id}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

#### Step 1 - Read the json file

```python
#define the schema
constructors_schema = """constructorId string,
                         name string,
                         nationality string,
                         url string"""
```

```python
#read data from constructors file
constructors_df = (
    spark.read
        .format('json')
        .schema(constructors_schema)
        .option('mode', 'FAILFAST')
        .load(source_file)
)
```

#### Step 2 - Add Metadata Columns

```python
constructors_final_df = add_ingestion_metadata(constructors_df)
```

#### Step 3 - Write to Bronze Delta Table

```python
write_to_bronze(input_df=constructors_final_df, target_table=table_name, batch_id=v_batch_id)
```
